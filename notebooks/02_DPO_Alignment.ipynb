```python
import json
import os
import shutil

cache_path = os.path.expanduser("~/.cache/huggingface/datasets")
if os.path.exists(cache_path):
    shutil.rmtree(cache_path)
    print("🧹 Cache cleared.")

input_file = "meme_expert_dpo_train_v1.jsonl" 
output_file = "fixed_swift_data.jsonl"

fixed_data = []

with open(input_file, 'r', encoding='utf-8') as f:
    content = f.read().replace('}{', '}\n{')

    for line in content.splitlines():
        line = line.strip()
        if not line: continue
        try:
            row = json.loads(line)
            new_row = {
                "messages": [
                    {"role": "system", "content": row.get("system", "")},
                    {"role": "user", "content": row["prompt"][0]["content"]},
                    {"role": "assistant", "content": row["chosen"][0]["content"]}
                ],
                "rejected_response": row["rejected"][0]["content"],
                "images": row.get("images", [])
            }
            fixed_data.append(new_row)
        except Exception as e:
            print(f"Skipping broken row: {e}")

with open(output_file, 'w', encoding='utf-8') as f:
    for entry in fixed_data:
        f.write(json.dumps(entry, ensure_ascii=False) + "\n")

print(f"✅ Created {output_file} with {len(fixed_data)} rows. Try training with this!")
```
